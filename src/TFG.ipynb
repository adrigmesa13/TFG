Contar datos originales

```python
%%sql
SELECT
    'listings' AS tabla,
    COUNT(*) AS registros_originales,
    COUNT(DISTINCT id) AS propiedades_unicas,
    COUNT(DISTINCT city) AS ciudades
FROM listings_all
UNION ALL
SELECT
    'reviews' AS tabla,
    COUNT(*) AS registros_originales,
    COUNT(DISTINCT listing_id) AS propiedades_unicas,
    COUNT(DISTINCT city) AS ciudades
FROM reviews_all
UNION ALL
SELECT
    'calendar' AS tabla,
    COUNT(*) AS registros_originales,
    COUNT(DISTINCT listing_id) AS propiedades_unicas,
    COUNT(DISTINCT city) AS ciudades
FROM calendar_all;
```

Contar nulos en listings (columnas críticas)

```python
%%sql
SELECT 'id' AS columna, COUNT(*) AS total, SUM(CASE WHEN id IS NULL THEN 1 ELSE 0 END) AS nulos, ROUND(100.0 * SUM(CASE WHEN id IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) AS pct_nulos FROM listings_all
UNION ALL SELECT 'city', COUNT(*), SUM(CASE WHEN city IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN city IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'property_type', COUNT(*), SUM(CASE WHEN property_type IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN property_type IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'room_type', COUNT(*), SUM(CASE WHEN room_type IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN room_type IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'price', COUNT(*), SUM(CASE WHEN price IS NULL OR price = '' THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN price IS NULL OR price = '' THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'bedrooms', COUNT(*), SUM(CASE WHEN bedrooms IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN bedrooms IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'accommodates', COUNT(*), SUM(CASE WHEN accommodates IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN accommodates IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'review_scores_rating', COUNT(*), SUM(CASE WHEN review_scores_rating IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN review_scores_rating IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'number_of_reviews', COUNT(*), SUM(CASE WHEN number_of_reviews IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN number_of_reviews IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
UNION ALL SELECT 'availability_365', COUNT(*), SUM(CASE WHEN availability_365 IS NULL THEN 1 ELSE 0 END), ROUND(100.0 * SUM(CASE WHEN availability_365 IS NULL THEN 1 ELSE 0 END) / COUNT(*), 2) FROM listings_all
ORDER BY pct_nulos DESC;
```

Crear tabla listings_clean

```python
%%sql
CREATE OR REPLACE TABLE listings_clean AS
SELECT
    id,
    city,
    property_type,
    room_type,
    accommodates,
    bedrooms,
    beds,
    bathrooms,
    CAST(REGEXP_REPLACE(REPLACE(price, ',', ''), '[^\d.]', '') AS DECIMAL(10,2)) AS price_clean,
    minimum_nights,
    maximum_nights,
    availability_365,
    number_of_reviews,
    reviews_per_month,
    review_scores_rating,
    review_scores_accuracy,
    review_scores_cleanliness,
    review_scores_checkin,
    review_scores_communication,
    review_scores_location,
    review_scores_value,
    host_id,
    host_is_superhost,
    host_listings_count,
    host_identity_verified,
    CASE WHEN picture_url IS NOT NULL THEN 1 ELSE 0 END AS has_picture_url,
    CASE WHEN amenities IS NOT NULL AND amenities != '' THEN 1 ELSE 0 END AS has_amenities
FROM listings_all
WHERE
    id IS NOT NULL
    AND city IS NOT NULL
    AND property_type IS NOT NULL
    AND room_type IS NOT NULL
    AND price IS NOT NULL AND price != '';
```

Crear tabla reviews_clean

```python
%%sql
CREATE OR REPLACE TABLE reviews_clean AS
SELECT
    id AS review_id,
    listing_id,
    DATE(date) AS review_date,
    city,
    CASE WHEN comments IS NOT NULL AND comments != '' THEN 1 ELSE 0 END AS has_comments
FROM reviews_all
WHERE
    id IS NOT NULL
    AND listing_id IS NOT NULL
    AND date IS NOT NULL
    AND city IS NOT NULL;
```

Crear tabla calendar_clean

```python
%%sql
CREATE OR REPLACE TABLE calendar_clean AS
SELECT
    listing_id,
    DATE(date) AS calendar_date,
    CASE WHEN available = 't' THEN 1 ELSE 0 END AS is_available,
    CAST(REGEXP_REPLACE(price, '[^\d.]', '') AS DECIMAL(10,2)) AS price_clean,
    minimum_nights,
    maximum_nights,
    city
FROM calendar_all
WHERE
    listing_id IS NOT NULL
    AND date IS NOT NULL
    AND available IS NOT NULL
    AND city IS NOT NULL;
```

estadisticas por ciudad

```python
%%sql
SELECT
    city,
    COUNT(DISTINCT id) AS total_propiedades,
    COUNT(DISTINCT host_id) AS total_hosts,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(MIN(price_clean), 2) AS precio_minimo,
    ROUND(MAX(price_clean), 2) AS precio_maximo,
    ROUND(STDDEV(price_clean), 2) AS precio_desv_std,
    ROUND(AVG(accommodates), 2) AS capacidad_promedio,
    ROUND(AVG(bedrooms), 2) AS dormitorios_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio
FROM listings_clean
GROUP BY city
ORDER BY total_propiedades DESC;
```

Distribución de ratings en rangos

```python
%%sql
SELECT
    city,
    CASE
        WHEN review_scores_rating >= 4.8 THEN '4.8-5.0 (Excelente)'
        WHEN review_scores_rating >= 4.5 THEN '4.5-4.79 (Muy Bueno)'
        WHEN review_scores_rating >= 4.0 THEN '4.0-4.49 (Bueno)'
        WHEN review_scores_rating >= 3.5 THEN '3.5-3.99 (Aceptable)'
        ELSE 'Menor 3.5 (Malo)'
    END AS rango_rating,
    COUNT(*) AS cantidad_propiedades,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio
FROM listings_clean
WHERE review_scores_rating IS NOT NULL
GROUP BY city, rango_rating
ORDER BY city, rango_rating DESC;
```

ANALIZAR RELACIONES ENTRE PRECIOS

Precios por type de propiedad

```python
%%sql
SELECT
    city,
    property_type,
    COUNT(*) AS cantidad,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(MIN(price_clean), 2) AS precio_min,
    ROUND(MAX(price_clean), 2) AS precio_max,
    ROUND(STDDEV(price_clean), 2) AS precio_desv_std,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio
FROM listings_clean
GROUP BY city, property_type
ORDER BY city, precio_promedio DESC;
```

Superhosts vs Hosts Normales - Precios y Rating

```python
%%sql
SELECT
    city,
    CASE WHEN host_is_superhost = 't' THEN 'Superhost' ELSE 'Host Normal' END AS tipo_host,
    COUNT(*) AS cantidad,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio,
    ROUND(AVG(reviews_per_month), 2) AS reviews_por_mes
FROM listings_clean
GROUP BY city, host_is_superhost
ORDER BY city, tipo_host DESC;
```

Precio vs Número de Dormitorios

```python
%%sql
SELECT
    city,
    bedrooms,
    COUNT(*) AS cantidad,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(MIN(price_clean), 2) AS precio_min,
    ROUND(MAX(price_clean), 2) AS precio_max,
    ROUND(AVG(accommodates), 2) AS capacidad_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio
FROM listings_clean
WHERE bedrooms IS NOT NULL
GROUP BY city, bedrooms
ORDER BY city, bedrooms;
```

Precio vs Capacidad de Huéspedes

```python
%%sql
SELECT
    city,
    accommodates,
    COUNT(*) AS cantidad,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(price_clean) / accommodates, 2) AS precio_por_persona,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio
FROM listings_clean
WHERE accommodates > 0
GROUP BY city, accommodates
ORDER BY city, accommodates;
```

Agrupar por rango de precio y analizar demanda

```python
%%sql
SELECT
    city,
    CASE
        WHEN price_clean < 100 THEN 'Muy Barato (<100€)'
        WHEN price_clean < 200 THEN 'Barato (100-200€)'
        WHEN price_clean < 350 THEN 'Medio (200-350€)'
        WHEN price_clean < 500 THEN 'Caro (350-500€)'
        WHEN price_clean < 750 THEN 'Muy Caro (500-750€)'
        ELSE 'Lujo (>750€)'
    END AS rango_precio,
    COUNT(*) AS cantidad_propiedades,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio,
    ROUND(AVG(reviews_per_month), 2) AS reviews_por_mes_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
    ROUND(AVG(accommodates), 2) AS capacidad_promedio
FROM listings_clean
GROUP BY city, rango_precio
ORDER BY city, rango_precio;
```

Variabilidad de precios

```python
%%sql
SELECT
    city,
    COUNT(*) AS cantidad,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(STDDEV(price_clean), 2) AS desv_std,
    ROUND(MIN(price_clean), 2) AS precio_min,
    ROUND(MAX(price_clean), 2) AS precio_max,
    ROUND(STDDEV(price_clean) / AVG(price_clean), 3) AS coef_variacion,
    ROUND(PERCENTILE_CONT(0.95) WITHIN GROUP (ORDER BY price_clean) - PERCENTILE_CONT(0.05) WITHIN GROUP (ORDER BY price_clean), 2) AS rango_80pct
FROM listings_clean
GROUP BY city
ORDER BY coef_variacion DESC;
```

OCUPACIÓN Y DEMANDA

Ocupación general por ciudad

```python
%%sql
SELECT
    city,
    COUNT(DISTINCT listing_id) AS propiedades_unicas,
    COUNT(*) AS total_dias_registrados,
    SUM(CASE WHEN is_available = 1 THEN 1 ELSE 0 END) AS dias_disponibles,
    SUM(CASE WHEN is_available = 0 THEN 1 ELSE 0 END) AS dias_ocupados,
    ROUND(100.0 * SUM(CASE WHEN is_available = 0 THEN 1 ELSE 0 END) / COUNT(*), 2) AS pct_ocupacion,
    ROUND(100.0 * SUM(CASE WHEN is_available = 1 THEN 1 ELSE 0 END) / COUNT(*), 2) AS pct_disponibilidad
FROM calendar_clean
GROUP BY city
ORDER BY pct_ocupacion DESC;
```

Clasificar propiedades por disponibilidad anual

```python
%%sql
SELECT
    city,
    CASE
        WHEN availability_365 > 300 THEN 'Siempre Disponible (>300 días)'
        WHEN availability_365 >= 200 THEN 'Muy Disponible (200-300 días)'
        WHEN availability_365 >= 100 THEN 'Parcialmente Disponible (100-200 días)'
        WHEN availability_365 >= 30 THEN 'Baja Disponibilidad (30-100 días)'
        ELSE 'Casi Nunca Disponible (<30 días)'
    END AS categoria_disponibilidad,
    COUNT(*) AS cantidad_propiedades,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio
FROM listings_clean
GROUP BY city, availability_365
ORDER BY city, availability_365 DESC;
```

Ocupación mensual por ciudad durante 2025

```python
%%sql
SELECT
    city,
    DATE_TRUNC('month', calendar_date)::DATE AS mes,
    EXTRACT(MONTH FROM calendar_date) AS numero_mes,
    COUNT(DISTINCT listing_id) AS propiedades_registradas,
    COUNT(*) AS total_dias,
    SUM(CASE WHEN is_available = 0 THEN 1 ELSE 0 END) AS dias_ocupados,
    SUM(CASE WHEN is_available = 1 THEN 1 ELSE 0 END) AS dias_disponibles,
    ROUND(100.0 * SUM(CASE WHEN is_available = 0 THEN 1 ELSE 0 END) / COUNT(*), 2) AS pct_ocupacion_mensual,
    ROUND(AVG(price_clean), 2) AS precio_promedio_mes
FROM calendar_clean
GROUP BY city, DATE_TRUNC('month', calendar_date), EXTRACT(MONTH FROM calendar_date)
ORDER BY city, mes;
```

Agrupar propiedades por demanda (número de reviews)

```python
%%sql
SELECT
    city,
    CASE
        WHEN number_of_reviews = 0 THEN 'Sin Reviews'
        WHEN number_of_reviews < 5 THEN 'Baja Demanda (1-5 reviews)'
        WHEN number_of_reviews < 20 THEN 'Demanda Media (5-20 reviews)'
        WHEN number_of_reviews < 50 THEN 'Alta Demanda (20-50 reviews)'
        ELSE 'Muy Alta Demanda (>50 reviews)'
    END AS categoria_demanda,
    COUNT(*) AS cantidad_propiedades,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(reviews_per_month), 2) AS reviews_por_mes,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
    ROUND(AVG(availability_365), 2) AS disponibilidad_promedio
FROM listings_clean
GROUP BY city, number_of_reviews
ORDER BY city, number_of_reviews;
```

Relación entre precio y ocupación

```python
%%sql
SELECT
    city,
    CASE
        WHEN price_clean < 150 THEN 'Barato (<150€)'
        WHEN price_clean < 300 THEN 'Medio (150-300€)'
        WHEN price_clean < 500 THEN 'Caro (300-500€)'
        ELSE 'Muy Caro (>500€)'
    END AS rango_precio,
    COUNT(*) AS cantidad,
    ROUND(AVG(availability_365), 2) AS disponibilidad_promedio,
    ROUND(100 - AVG(availability_365), 2) AS ocupacion_promedio,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio,
    ROUND(AVG(reviews_per_month), 2) AS reviews_por_mes
FROM listings_clean
GROUP BY city, rango_precio
ORDER BY city, rango_precio;
```

SATISFACCIÓN Y RATINGS

Rating general por ciudad y tipo de propiedad

```python
%%sql
SELECT
    city,
    room_type,
    COUNT(*) AS cantidad_propiedades,
    ROUND(AVG(review_scores_rating), 2) AS rating_general,
    ROUND(AVG(review_scores_cleanliness), 2) AS rating_limpieza,
    ROUND(AVG(review_scores_communication), 2) AS rating_comunicacion,
    ROUND(AVG(review_scores_accuracy), 2) AS rating_precision,
    ROUND(AVG(review_scores_location), 2) AS rating_ubicacion,
    ROUND(AVG(review_scores_value), 2) AS rating_relacion_precio,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio
FROM listings_clean
WHERE review_scores_rating IS NOT NULL
GROUP BY city, room_type
ORDER BY city, rating_general DESC;
```

Dimensión más débil por ciudad

```python
%%sql
SELECT
    city,
    COUNT(*) AS total_propiedades,
    ROUND(AVG(review_scores_rating), 2) AS rating_general,
    ROUND(AVG(review_scores_cleanliness), 2) AS limpieza,
    ROUND(AVG(review_scores_communication), 2) AS comunicacion,
    ROUND(AVG(review_scores_accuracy), 2) AS precision,
    ROUND(AVG(review_scores_location), 2) AS ubicacion,
    ROUND(AVG(review_scores_value), 2) AS valor,
    ROUND(AVG(review_scores_checkin), 2) AS checkin,
    LEAST(
        ROUND(AVG(review_scores_cleanliness), 2),
        ROUND(AVG(review_scores_communication), 2),
        ROUND(AVG(review_scores_accuracy), 2),
        ROUND(AVG(review_scores_location), 2),
        ROUND(AVG(review_scores_value), 2),
        ROUND(AVG(review_scores_checkin), 2)
    ) AS puntuacion_minima
FROM listings_clean
WHERE review_scores_rating IS NOT NULL
GROUP BY city
ORDER BY city;
```

Superhosts vs Hosts Normales - Desglose de ratings

```python
%%sql
SELECT
    city,
    CASE WHEN host_is_superhost = 't' THEN 'Superhost' ELSE 'Host Normal' END AS tipo_host,
    COUNT(*) AS cantidad,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(review_scores_rating), 2) AS rating_general,
    ROUND(AVG(review_scores_cleanliness), 2) AS rating_limpieza,
    ROUND(AVG(review_scores_communication), 2) AS rating_comunicacion,
    ROUND(AVG(review_scores_accuracy), 2) AS rating_precision,
    ROUND(AVG(review_scores_location), 2) AS rating_ubicacion
FROM listings_clean
WHERE review_scores_rating IS NOT NULL
GROUP BY city, host_is_superhost
ORDER BY city, tipo_host DESC;
```

Distribución de ratings en rangos

```python
%%sql
SELECT
    city,
    CASE
        WHEN review_scores_rating >= 4.9 THEN '4.9-5.0 (Perfecto)'
        WHEN review_scores_rating >= 4.8 THEN '4.8-4.89 (Excelente)'
        WHEN review_scores_rating >= 4.5 THEN '4.5-4.79 (Muy Bueno)'
        WHEN review_scores_rating >= 4.0 THEN '4.0-4.49 (Bueno)'
        WHEN review_scores_rating >= 3.5 THEN '3.5-3.99 (Aceptable)'
        ELSE '<3.5 (Malo)'
    END AS rango_rating,
    COUNT(*) AS cantidad_propiedades,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio
FROM listings_clean
WHERE review_scores_rating IS NOT NULL
GROUP BY city, rango_rating
ORDER BY city, rango_rating DESC;
```

Relación entre actividad (reviews) y satisfacción

```python
%%sql
SELECT
    city,
    CASE
        WHEN number_of_reviews = 0 THEN 'Sin Reviews'
        WHEN number_of_reviews < 5 THEN 'Baja Actividad (1-5)'
        WHEN number_of_reviews < 20 THEN 'Media Actividad (5-20)'
        WHEN number_of_reviews < 50 THEN 'Alta Actividad (20-50)'
        ELSE 'Muy Alta Actividad (>50)'
    END AS actividad,
    COUNT(*) AS cantidad,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
    ROUND(AVG(reviews_per_month), 2) AS reviews_por_mes
FROM listings_clean
GROUP BY city, number_of_reviews
ORDER BY city, number_of_reviews;
```

COMPARATIVA ENTRE CIUDADES

Tabla resumen completa de todas las ciudades

```python
%%sql
SELECT
    city,
    COUNT(DISTINCT id) AS total_propiedades,
    COUNT(DISTINCT host_id) AS total_hosts,
    ROUND(COUNT(DISTINCT id)::NUMERIC / COUNT(DISTINCT host_id), 2) AS propiedades_por_host,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(MIN(price_clean), 2) AS precio_minimo,
    ROUND(MAX(price_clean), 2) AS precio_maximo,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
    ROUND(AVG(accommodates), 2) AS capacidad_promedio,
    ROUND(AVG(bedrooms), 2) AS dormitorios_promedio,
    COUNT(CASE WHEN host_is_superhost = 't' THEN 1 END) AS superhosts,
    ROUND(100.0 * COUNT(CASE WHEN host_is_superhost = 't' THEN 1 END) / COUNT(DISTINCT host_id), 2) AS pct_superhosts,
    ROUND(AVG(availability_365), 2) AS disponibilidad_promedio,
    ROUND(100 - AVG(availability_365), 2) AS ocupacion_promedio,
    ROUND(AVG(number_of_reviews), 2) AS reviews_promedio,
    ROUND(AVG(reviews_per_month), 2) AS reviews_por_mes
FROM listings_clean
GROUP BY city
ORDER BY total_propiedades DESC;
```

Concentración del mercado (propiedades por host)

```python
%%sql
SELECT
    city,
    COUNT(DISTINCT host_id) AS total_hosts,
    COUNT(DISTINCT id) AS total_propiedades,
    ROUND(COUNT(DISTINCT id)::NUMERIC / COUNT(DISTINCT host_id), 2) AS propiedades_por_host_promedio,
    ROUND(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY host_listings_count), 2) AS mediana_propiedades_por_host,
    MAX(host_listings_count) AS max_propiedades_un_host,
    COUNT(CASE WHEN host_listings_count > 5 THEN 1 END) AS hosts_con_mas_de_5,
    COUNT(CASE WHEN host_listings_count = 1 THEN 1 END) AS hosts_con_una_sola
FROM listings_clean
GROUP BY city
ORDER BY propiedades_por_host_promedio DESC;
```

Especialización por tipo de propiedad

```python
%%sql
SELECT
    city,
    room_type,
    COUNT(*) AS cantidad,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (PARTITION BY city), 2) AS pct_del_total,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio
FROM listings_clean
GROUP BY city, room_type
ORDER BY city, pct_del_total DESC;
```

Ranking de ciudades por diferentes métricas

```python
%%sql
WITH city_metrics AS (
    SELECT
        city,
        COUNT(DISTINCT id) AS total_propiedades,
        ROUND(AVG(price_clean), 2) AS precio_promedio,
        ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
        ROUND(100 - AVG(availability_365), 2) AS ocupacion_promedio
    FROM listings_clean
    GROUP BY city
)
SELECT
    city,
    total_propiedades,
    RANK() OVER (ORDER BY total_propiedades DESC) AS rank_propiedades,
    precio_promedio,
    RANK() OVER (ORDER BY precio_promedio DESC) AS rank_precio_alto,
    rating_promedio,
    RANK() OVER (ORDER BY rating_promedio DESC) AS rank_rating,
    ocupacion_promedio,
    RANK() OVER (ORDER BY ocupacion_promedio DESC) AS rank_ocupacion
FROM city_metrics
ORDER BY total_propiedades DESC;
```

Clasificación de ciudades por madurez de mercado

```python
%%sql
SELECT
    city,
    COUNT(DISTINCT id) AS total_propiedades,
    CASE
        WHEN COUNT(DISTINCT id) > 15000 THEN 'Mercado Muy Grande'
        WHEN COUNT(DISTINCT id) >= 7000 THEN 'Mercado Grande'
        WHEN COUNT(DISTINCT id) >= 3000 THEN 'Mercado Mediano'
        ELSE 'Mercado Pequeño'
    END AS tamaño_mercado,
    ROUND(AVG(price_clean), 2) AS precio_promedio,
    ROUND(AVG(review_scores_rating), 2) AS rating_promedio,
    ROUND(100 - AVG(availability_365), 2) AS ocupacion_promedio,
    ROUND(COUNT(DISTINCT id)::NUMERIC / COUNT(DISTINCT host_id), 2) AS propiedades_por_host
FROM listings_clean
GROUP BY city
ORDER BY total_propiedades DESC;
```
